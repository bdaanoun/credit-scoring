```python
import pandas as pd

df = pd.read_csv("../data/application_train.csv")
df
```

```python
chunk_size = 10
chunks = {f"Cols {i+1}-{i+chunk_size}": df.columns[i:i+chunk_size].tolist() 
          for i in range(0, len(df.columns), chunk_size)}

# Fills missing values automatically if the last chunk is smaller
grid_df = pd.DataFrame.from_dict(chunks, orient='index').transpose().fillna("")
display(grid_df)
```
